```python
import plotly.graph_objects as go

months   = ['Aug','Sep','Oct','Nov','Dec','Jan','Feb']
baseline = [42000,44000,43000,44200,43500,44000,43800]
with_eng = [42000,44000,43000,47800,51200,54100,55900]

fig = go.Figure()
fig.add_trace(go.Scatter(x=months, y=baseline,
    name='Without Pricing Engine', line=dict(color='#ff4d00', dash='dash')))
fig.add_trace(go.Scatter(x=months, y=with_eng,
    name='With Pricing Engine', line=dict(color='#00875a'),
    fill='tonexty', fillcolor='rgba(0,135,90,0.08)'))
fig.update_layout(title='A/B Revenue Simulation: Engine vs No Engine')
fig.show()
```
